```python
file_path = "imadole_cricket_club_stats.csv"
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(file_path)
print(df.head())      # check first few rows
print(df.columns)     # check column names
```

```python
df = df["Player,Runs,Strike_Rate,Economy_Rate"].str.split(",", expand=True)
df.columns = ["Player", "Runs", "Strike_Rate", "Economy_Rate"]
df["Runs"] = pd.to_numeric(df["Runs"])  # convert Runs to numbers
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Runs"], bins=8, color='skyblue', edgecolor='black')
plt.title("Figure 1: Histogram of Runs Scored")
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Runs"], df["Strike_Rate"])
plt.title("Figure 2: Runs vs Strike Rate")
plt.xlabel("Runs")
plt.ylabel("Strike Rate")
plt.show()
```

```python
print(df["Economy_Rate"].dtype)
```

```python
df["Economy_Rate"] = pd.to_numeric(df["Economy_Rate"], errors='coerce')
```

```python
df = df.dropna(subset=["Economy_Rate"])
```

```python
plt.figure()
plt.boxplot(df["Economy_Rate"])
plt.title("Figure 3: Economy Rate Distribution")
plt.ylabel("Economy Rate")
plt.show()
```
